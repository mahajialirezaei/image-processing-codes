<p dir=rtl style="direction: rtl;text-align: justify;line-height:200%;font-family:vazir;font-size:medium">

الگوریتم کلی تابع به این شکل هست که شما باید ابتدا خود پلاک را تشخیص دهید. سپس بعد از مطمئن شدن از درستی الگوریتم تشخیص پلاک خود، سعی کنید کاراکترها را پیدا کنید. سپس مجموعه‌ای از اعداد 0-9 را بصورت تصویر در پوشه با مسیر numbers ذخیره کنید و کاراکترهای پیدا شده را با تصاویر موجود در پوشه numbers که خود ساخته‌اید مقایسه کنید تا عدد استخراج شود.

<p dir=rtl style="direction: rtl;text-align: justify;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazir" size=3>
1- چه به کمک کد پایتون چه با استفاده از ابزارهای ویندوز، پوشه numbers را در همین مسیر بسازید و با استفاده از داده‌هایی که در اختیارتان قرار گرفته اعداد 0-9 را جدا کرده و از هر رقم حداقل یک تصویر در دایرکتوری خود ذخیره داشته باشید. و دقت کنید که نام تصویر ذخیره شده عدد مورد نظر باشد برای مثال 2.png در نهایت در نظر داشته باشید که شما قرار هست کاراکترهای روی پلاک را تشخیص داده و با استفاده از تصویر اعداد ذخیره شده، متوجه شوید هر کاراکتر تشخیص داده شده چه عددی است.
<br>
2- برگردیم سر نوشتن تابع! عملیات لبه‌یابی را روی تصاویری که در اختیارتان قرار گرفته انجام دهید. تا لبه‌های پلاک تیز تر شود. انتخاب پارامترهای مناسب با شما!
<br>
3- در این مرحله شما باید پلاک را تشخیص دهید. با استفاده از تابع fin_contour این اتفاق شدنی‌ست. فقط باید با توجه به تصاویری که در اختیار دارید. ابعاد کانتور را طوری طراحی کنید تا فقط پلاک پیدا شود و بقیه کانتورهای حاضر در تصویر نادید گرفته شوند. شما باید دو شرط را رعایت کنید. نخست مستطیل بودن کانتور و سپس ابعاد کانتور که با سعی و خطا روی داده‌ها و تصاویر موجود بدست می‌آید. دقت کنید تصاویر اولیه که در اختیار دارید، کلید حل مسئله است.
<br>
4- در این مرحله ما فرض می‌کنیم که مستطیل پلاک پیدا شده. دوباره دقت کنید که ما فقط به 5 رقم اول نیاز داریم پس اگر فقط همان 5 عدد به عنوان کانتور تشخیص داده شد، جای نگرانی نیست. حال باید کاراکترهای داخل پلاک پیدا شوند. برای این کار دوباره از همان تابع find_contours استفاده می‌کنیم و ابعاد را طوری تنظیم می‌کنیم که کاراکترها را پیدا کند. دوباره باید دقت کنید که کانتور به شکل چهار ضلعی باشد و ابعاد خاصی داشته باشد که آن ابعاد با استفاده از سعی و خطا و نگاه به تصاویری که دارید، بدست می‌آید.
<br>
5- تصاویر پوشه numbers را بارگذاری کنید.
<br>
6- حالا کاراکترهای پیدا شده را با تصاویر موجود در پوشه numbers شباهت سنجی کنید. از روش اختلاف مربعات پیکسل‌ها (sum of squared differences) برای محاسبه شباهت استفاده کنید. برای هر کاراکتر، قالبی که کمترین اختلاف را دارد به‌عنوان رقم شناسایی‌شده انتخاب کنید.
<br>
7- ارقام پیدا شده. حالا کافی است میانگین حساب کنید.
</font>
</p>

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


def process_plate(image):
        def find_plate(image):
            if len(image.shape) == 3:
                gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
            else:
                gray = image.copy()

            _, binary_img = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

            res = cv2.findContours(binary_img, cv2.RETR_LIST, cv2.CHAIN_APPROX_SIMPLE)
            contours = res[0] if len(res) == 2 else res[1]
            max_a = 0
            cnt_find = []
            for cnt in contours:

                area = cv2.contourArea(cnt)
                peri = cv2.arcLength(cnt, True)
                approx = cv2.approxPolyDP(cnt, 0.02 * peri, True)
                if (peri/4)**2 == area:
                    continue

                x, y, w, h = cv2.boundingRect(approx)

                if not (20< w<= 400 and 20<h <= 105):
                    continue

                print("bbox:", x, y, w, h)
                if area > max_a:
                    cnt_find= cnt
                    max_a = area
                # break

            if cnt_find is None:
                print("No contour found.")
                return None

            vis = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)
            cv2.drawContours(vis, [cnt_find], -1, (0,255,0), 2)
            plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB))


            return cnt_find
        plate = find_plate(image)


process_plate(image=img)
```
